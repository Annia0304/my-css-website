4) Palindrome

```python
def is_palindrome(text):
  text = text.replace(" ", "").lower()
  return text == text[::-1]

user_string = input("Enter a string to check for palindrome: ")

if is_palindrome(user_string):
  print(f"'{user_string}' is a palindrome.")
else:
  print(f"'{user_string}' is not a palindrome.")
```

5) Program to upload a image and perform tasks

```python
# Step 1: Install libraries
!pip install opencv-python matplotlib pillow

# Step 2: Import required libraries
from google.colab import files
import cv2
import matplotlib.pyplot as plt

# Step 3: Upload image
uploaded = files.upload()

# Step 4: Load uploaded image
for fn in uploaded.keys():
    img_path = fn
    img = cv2.imread(img_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR -> RGB

# Step 5: Display original image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()

# ---- Task 1: Convert to Grayscale ----
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

# ---- Task 2: Gaussian Blur ----
blur = cv2.GaussianBlur(img_rgb, (15, 15), 0)
plt.imshow(blur)
plt.title("Blurred Image")
plt.axis("off")
plt.show()

# ---- Task 3: Edge Detection ----
edges = cv2.Canny(img, 100, 200)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis("off")
plt.show()
```

6) Phyton program to handle files

```python
# prompt: uploading the pdf tell amount of characters

!pip install PyPDF2
from google.colab import files
import io
import PyPDF2

uploaded = files.upload()

for filename in uploaded.keys():
    print(f'User uploaded file "{filename}" with length {len(uploaded[filename])} bytes')

    # Read the PDF file from bytes
    pdf_reader = PyPDF2.PdfReader(io.BytesIO(uploaded[filename]))
    text = ""
    for page_num in range(len(pdf_reader.pages)):
        page = pdf_reader.pages[page_num]
        text += page.extract_text()

    character_count = len(text)

    print(f"The total number of characters in the PDF is: {character_count}")
```
